## Historical research priorities

The project moved from surrogate-model iteration to parent reconstruction, acquisition and validation auditing, a negative geometry screen, and qualification of a pretrained knee-anatomy route.

The following **planning priorities are historical ordinal scores**, preserved from this notebook's original source. They are not accuracy, measured performance, probabilities of success, or a current experiment ranking.


```python
from pathlib import Path
import io
import json

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import plotly.graph_objects as go
from IPython.display import display

ROOT = next(
    path for path in (Path.cwd(), *Path.cwd().parents)
    if (path / "reports/current_frontier/results.json").is_file()
)
report = json.loads((ROOT / "reports/current_frontier/results.json").read_text())


def display_chart(fig, static):
    """Persist interactive Plotly and static SVG from the same chart variables."""
    fig.update_layout(
        template="plotly_white", width=900, height=440,
        margin=dict(l=85, r=35, t=80, b=80), font=dict(size=14),
    )
    buffer = io.StringIO()
    static.savefig(buffer, format="svg", bbox_inches="tight")
    plt.close(static)
    display({
        "application/vnd.plotly.v1+json": json.loads(fig.to_json()),
        "image/svg+xml": buffer.getvalue(),
        "text/plain": fig.layout.title.text,
    }, raw=True)

# Historical planning scores from the original published notebook, not metrics.
priorities = {
    'E87A': 89, 'E87B': 87, 'E87C': 81,
    'ANATOMY_PILOT': 80, 'E87F': 77, 'E87E': 71,
}
labels = list(priorities)
values = list(priorities.values())
fig = go.Figure(go.Bar(x=labels, y=values, marker_color="#155e75"))
fig.update_layout(
    title="Historical research priorities — planning scores",
    yaxis_title="Historical planning score (not model performance)",
    yaxis_range=[0, 100],
)
static, ax = plt.subplots(figsize=(10, 4))
ax.bar(labels, values, color="#155e75")
ax.set(
    ylim=(0, 100), ylabel="Historical planning score",
    title="Historical research priorities — planning scores",
)
for i, value in enumerate(values):
    ax.text(i, value + 1, str(value), ha="center")
display_chart(fig, static)
print({
    'external_macro_roc_auc': report['public_score']['verified_official_score'],
    'primary_direction': 'anatomy_localized_global_local_parent_residual',
    'priorities': priorities, 'scope': 'historical Stage 96 planning snapshot',
})
```

## Verified engineering evidence

Each branch or audit has a separate acceptance contract. Counts below use different units and must not be summed into a single model-quality or completion score. The first five counts load directly from the published aggregate report; the four anatomy qualification tracks are the historical Stage 96 record preserved in this notebook and [qualification documentation](../docs/ANATOMY_MODEL_QUALIFICATION.md).

Complete parent parity and untouched fully labeled confirmation are not claimed where source lineage does not support them. The physical geometry audit inspected **2,287 headers** and found **zero geometry flags** in its scoped inputs; that result does not establish complete acquisition coverage.


```python
parent = report['parent_reconstruction']
audit = report['stage93_validation_acquisition_audit']
geometry = report['stage95_geometry_audit']
completed = {
    'Native members': parent['native_members_executed'],
    'A5 folds': parent['a5_folds_executed'],
    'Raptor views': parent['raptor_diagnostic_views_completed'],
    'CoAt predictions': parent['coat_trained_predictions_completed'],
    'Audit units': audit['units_completed'],
    'Anatomy tracks': 4,  # Historical Stage 96 qualification, not segmentation quality.
}
labels = list(completed)
values = list(completed.values())
fig = go.Figure(go.Bar(x=labels, y=values, marker_color="#155e75"))
fig.update_layout(
    title="Verified engineering units by milestone — through Stage 96",
    yaxis_title="Completed units (different units by milestone)",
)
static, ax = plt.subplots(figsize=(11, 4))
ax.bar(labels, values, color="#155e75")
ax.set(
    ylim=(0, max(values) + 4), ylabel="Completed units",
    title="Verified engineering units by milestone — through Stage 96",
)
for i, value in enumerate(values):
    ax.text(i, value + 0.3, str(value), ha="center")
display_chart(fig, static)
print({
    'completed': completed,
    'geometry_headers_inspected': geometry['headers_inspected'],
    'geometry_flags': geometry['geometry_flags'],
    'anatomy_lifecycle': 'QUALIFIED_FOR_REFERENCE_PILOT',
    'scope': 'historical Stage 96 qualification, not current anatomy status',
})
```
