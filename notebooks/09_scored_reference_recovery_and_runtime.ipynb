## Trained Rad-head optimization experiment

A source-derived optimization removes five diagnostic-only head-prediction passes while retaining the contributing heads, calibration, and ranking path. Stage 63 loaded **15 trained heads** and measured three GPU trials using synthetic encoder features. Final values and CSV bytes were exact in every trial.

The measured **1.274× ratio is head-only**. Encoder preprocessing, real-image parity, and end-to-end submission runtime are not yet verified, so the optimization is not activated as a production claim.

```python
baseline=[0.02673059500011732,0.026663292000193906,0.0265207500001452]
candidate=[0.02350604199978079,0.020931547000145656,0.02078682299952561]
timing=pd.DataFrame({"Trial":[1,2,3]*2,"Path":["Original"]*3+["Candidate"]*3,"Head-only seconds":baseline+candidate})
fig=px.bar(timing,x="Trial",y="Head-only seconds",color="Path",barmode="group",title="Trained Rad-head timing: original vs candidate")
fig.update_layout(height=430)
fig.show()
print("SCORED_REFERENCE_FRONTIER_PUBLICATION_COMPLETE")
```
