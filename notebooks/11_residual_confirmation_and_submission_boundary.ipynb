# 11 — Residual confirmation and submission boundary

This notebook publishes aggregate-only evidence from the Stage 70 screen and Stage 71 independent confirmation. The historical **0.933** Kaggle score remains the official control; internal grouped AUC is not treated as directly comparable to it.

Private MRI, identifiers, row-level predictions, checkpoints, sampling schedules, credentials, and submission glue are intentionally excluded.

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
stages=['Stage 70 screen','Stage 71 confirmation']
control=[0.7878272932063983,0.7979362002501319]
blend=[0.7891558134890007,0.8003733592238084]
fig=go.Figure()
fig.add_bar(name='Grouped control',x=stages,y=control)
fig.add_bar(name='Fixed residual blend',x=stages,y=blend)
fig.update_layout(title='Grouped control versus frozen residual blend',barmode='group',height=430,yaxis_title='Macro ROC-AUC',yaxis_range=[0.75,0.82])
fig.show()
```

## Independent confirmation

Stage 70 passed its frozen screening gates and advanced unchanged to folds 1, 3, and 4. Stage 71 improved the grouped mean, but one confirmation fold was marginally negative. Because the decision rule required every confirmation fold to be nonnegative, the branch was not promoted.

```python
folds=pd.DataFrame({'Fold':['Fold 1','Fold 3','Fold 4'],'Delta':[0.002299888073846512,-0.00008538643241529087,0.005096975279598404]})
fig=px.bar(folds,x='Fold',y='Delta',text='Delta',title='Stage 71 fixed-blend delta by confirmation fold')
fig.update_traces(texttemplate='%{text:.6f}',textposition='outside')
fig.add_hline(y=0,line_dash='dash')
fig.update_layout(height=430,yaxis_title='Δ macro ROC-AUC')
fig.show()
```

## Frozen gate outcome

Five confirmation gates passed: candidate floor, mean blend gain, worst-target regression limit, bootstrap-positive fraction, and diversity. The only failed gate was the all-folds-nonnegative rule. This is recorded as a valid scientific negative result rather than weakening the rule after seeing the outcome.

Stage 72 then moved to submission-boundary engineering. No new official Kaggle score is claimed in this notebook.

```python
gates=pd.DataFrame({'Gate':['candidate floor','mean blend gain','all folds nonnegative','worst target','bootstrap','diversity'],'Pass':[1,1,0,1,1,1]})
fig=px.bar(gates,x='Gate',y='Pass',text=gates['Pass'].map({1:'PASS',0:'FAIL'}),title='Stage 71 preregistered confirmation gates')
fig.update_traces(textposition='outside')
fig.update_layout(height=430,yaxis_range=[0,1.2],yaxis_tickvals=[0,1],yaxis_ticktext=['Fail','Pass'])
fig.show()
print('RESIDUAL_CONFIRMATION_PUBLICATION_COMPLETE')
```
