```python
from pathlib import Path
import json
import sys
ROOT = Path.home() / "rsna-knee-abnormality-detection"
assert (ROOT / "configs/milestone01.json").is_file(), "Open the installed project, not a loose notebook."
assert Path(sys.prefix).resolve() == (ROOT / ".venv").resolve(), "Select kernel: RSNA Knee - audit"
sys.path.insert(0, str(ROOT / "src"))
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
from IPython.display import display, Markdown
from rsna_knee.runtime import require_tests
pio.renderers.default = "plotly_mimetype"
TESTS = require_tests(ROOT)
print(f"Matching synthetic test receipt: {TESTS['tests']} tests; failures={TESTS['failures']}; errors={TESTS['errors']}")
print("No project model has been fitted by this notebook.")
```

```python
from rsna_knee.runtime import inventory
PREFLIGHT = inventory(ROOT)
print("Preflight resource gate passed; no AWS account inspection performed.")
print(f"Logical CPUs visible to process: {PREFLIGHT['logical_cpus']}")
print(f"Python: {PREFLIGHT['python']}")
print(f"Source fingerprint: {PREFLIGHT['source_hash']}")
```

### Memory available to this environment
The audit worker has a 4 GiB resident-memory guard. This chart is a point-in-time system reading, not a promise that other kernels will leave that memory free. Close unnecessary notebooks rather than immediately upgrading hardware.


```python
memory = pd.DataFrame({"state": ["Available", "Other memory"],
    "GiB": [PREFLIGHT["available_ram_gib"], max(0, PREFLIGHT["total_ram_gib"] - PREFLIGHT["available_ram_gib"])]})
fig = px.bar(memory, x="state", y="GiB", title="Observed RAM availability — preflight", text_auto=".2f")
fig.update_layout(xaxis_title="Memory state", yaxis_title="GiB", height=380)
fig.show(renderer="plotly_mimetype")
```

### Disk headroom
Only five small metadata CSVs are requested in this milestone. **Do not download the full MRI archive.** Disk readings can differ from the volume size selected in AWS; confirm that setting in the space details rather than inferring a bill from this plot.


```python
disk = pd.DataFrame({"state": ["Used", "Free"], "GiB": [PREFLIGHT["disk_used_gib"], PREFLIGHT["disk_free_gib"]]})
fig = px.bar(disk, x="state", y="GiB", title="Observed project-filesystem capacity", text_auto=".2f")
fig.update_layout(xaxis_title="Disk state", yaxis_title="GiB", height=380)
fig.show(renderer="plotly_mimetype")
```

### Reproducible environment evidence
Direct dependency versions are pinned in `requirements-audit.txt`; the full dependency resolution has not been pre-validated by the assistant. The receipt records what actually installed. A mismatch between source/environment and test evidence stops subsequent stages.


```python
packages = pd.DataFrame(list(PREFLIGHT["versions"].items()), columns=["Package", "Observed version"])
fig = go.Figure(data=[go.Table(header={"values": ["Package", "Observed version"]},
    cells={"values": [packages["Package"], packages["Observed version"]]})])
fig.update_layout(title="Installed audit environment — observed versions", height=460)
fig.show(renderer="plotly_mimetype")
```
